# 2f4a05.04-推理pipeline

`pipeline()` 让使用[Hub](https://huggingface.co/models)上的任何模型进行任何语言、计算机视觉、语音以及多模态任务的推理变得非常简单。即使您对特定的模态没有经验，或者不熟悉模型的源码，您仍然可以使用`pipeline()`进行推理！

本节将介绍：

*   如何使用`pipeline()` 进行推理。
*   如何使用特定的`tokenizer`(分词器)或模型。
*   如何使用`pipeline()` 进行音频、视觉和多模态任务的推理。

可以自行搜索查看`pipeline()`文档以获取已支持的任务和可用参数的完整列表。

## Pipeline使用

- 虽然每个任务都有一个关联的`pipeline()`，但使用通用的抽象的`pipeline()`更加简单，
- 其中包含所有特定任务的`pipelines`。
- `pipeline()`会自动加载一个默认模型和一个能够进行任务推理的预处理类。
- 举例：以使用`pipeline()`进行自动语音识别（ASR）或语音转文本为例。

1.  首先，创建一个`pipeline()`并指定推理任务：

```python
from transformers import pipeline

transcriber = pipeline(task="automatic-speech-recognition")
```

**`pipeline()`的在线运行机制**
- 当第一次运行这段代码时，`pipeline` 会根据任务类型（在此例中为自动语音识别）下载相应的模型。
- 这一过程是自动完成的，用户只需提供任务名称和所需的输入数据（如音频文件的链接）
- `pipeline()`就可以直接进行语音识别，而无需额外的下载步骤

2.  将您的输入传递给`pipeline()`。对于语音识别，这通常是一个音频输入文件：

**马丁路德金音频：I have a dream... ** 可以在文件夹打开倾听

## 参数

- `pipeline()` 支持许多参数；
- 有些是适用于特定任务的，而有些适用于所有`pipeline`。
- 通常情况下，您可以在任何地方指定对应参数：

```python
transcriber = pipeline(model="openai/whisper-large-v2", my_parameter=1)

out = transcriber(...)  # This will use `my_parameter=1`.
out = transcriber(..., my_parameter=2)  # This will override and use `my_parameter=2`.
out = transcriber(...)  # This will go back to using `my_parameter=1`.
```

## pipeline()其中的三个重要参数：

### 设备(对于无GPU，或者单GPU，不用看)

- 如果您使用 `device=n`，`pipeline`会自动将模型放在指定的设备n上。
- 无论使用PyTorch还是Tensorflow，这都可以工作。

```python
transcriber = pipeline(model="openai/whisper-large-v2", device=0)
```

- 如果模型对于单个GPU来说过于庞大，并且您正在使用PyTorch，
- 可以设置 `device_map="auto"` 以自动确定如何加载和存储模型权重。
- <mark>使用 `device_map` 参数需要安装🤗 [Accelerate](https://huggingface.co/docs/accelerate) 软件包</mark>：

```python
pip install --upgrade accelerate
```

- Accelerate 是一个 Python 库，可以帮助你在多 GPU 或 TPU 上运行 PyTorch 代码，而无需修改你的原始代码。
- 它是由 Hugging Face 公司开发的，旨在提供一个高性能、高能效的计算平台，用于机器学习、自然语言处理和其他领域。

- Accelerate 的一个主要特点是 device_map 参数，
- 它可以让你指定如何在不同的设备上加载和存储模型权重。
- 你可以设置 device_map=“auto”，让 Accelerate 自动检测可用的设备，并根据你的模型大小和设备数量，分配最佳的权重分布方案。
- 这样，你就不需要手动管理多个模型副本或使用复杂的分布式训练框架了。

### 实例
- 你可以使用 Hugging Face 的 pipeline 函数来创建一个转录器，它可以将语音转换为文本。
- 你只需要指定模型的名称和 device_map=“auto”，就可以让转录器在所有 GPU 设备上运行，提高转录速度和准确度。
- 以下是一个示例代码：

from transformers import pipeline
transcriber = pipeline(model="openai/whisper-large-v2", device_map="auto")

- 这样，就可以使用 transcriber 对任何语音文件进行转录，而不需要关心设备的配置和管理。

借助accelerate + device_map="auto"

以下代码会自动在所有GPU设备(多GPU卡)上加载和存储模型权重：

```python
transcriber = pipeline(model="openai/whisper-large-v2", device_map="auto")
```

请注意，如果传递了 `device_map="auto"`，就不需要再添加 `device=device` 参数，否则可能会遇到一些意外的状况！

### 批量大小（了解）

- 默认情况下，`pipelines`不会进行批量推理，原因在[这里](https://huggingface.co/docs/transformers/main_classes/pipelines#pipeline-batching)详细解释。
- 因为批处理不一定更快，实际上在某些情况下可能会更慢。

但如果在您的用例中要求使用Batch：

```python
transcriber = pipeline(model="openai/whisper-large-v2", device=0, batch_size=2)
audio_filenames = [f"https://huggingface.co/datasets/Narsil/asr_dummy/resolve/main/{i}.flac" for i in range(1, 5)]
texts = transcriber(audio_filenames)
```

- 以上代码会在提供的4个音频文件上运行`pipeline`，
- 它会将它们以2个一组的批次传递给模型（模型在GPU上，此时批处理更有可能有所帮助），而您无需编写额外的代码。
- 输出应始终与没有批处理时收到的结果相一致。
- Batch只是一种帮助您更快地使用`pipeline`的方式。

- `pipeline`也可以减轻一些批处理的复杂性，因为对于某些`pipeline`，需要将单个项目（如长音频文件）分成多个部分以供模型处理。
- `pipeline`为您执行这种[_chunk batching_](https://huggingface.co/docs/transformers/v4.35.0/zh/main_classes/pipelines#pipeline-chunk-batching)。

### 任务特定参数

- pipeline里面可以加载不同类型的任务
- 各种任务都提供了特定于任务的参数，这些参数随不同任务而各自定义
- 这些参数提供额外的灵活性和选择，以帮助您完成工作。
- 例如，`transformers.AutomaticSpeechRecognitionPipeline.__call__()` 方法具有一个 `return_timestamps` 参数，对于字幕视频似乎很有帮助：

```python
transcriber = pipeline(model="openai/whisper-large-v2", return_timestamps=True)
transcriber("https://huggingface.co/datasets/Narsil/asr_dummy/resolve/main/mlk.flac")
{'text': ' I have a dream that one day this nation will rise up and live out the true meaning of its creed.', 'chunks': [{'timestamp': (0.0, 11.88), 'text': ' I have a dream that one day this nation will rise up and live out the true meaning of its'}, {'timestamp': (11.88, 12.38), 'text': ' creed.'}]}
```

正如您所看到的，模型推断出了文本，还输出了各个句子发音的**时间**。

- 每个任务都有许多可用的参数，
- 因此请查看每个任务的API参考，以了解您可以进行哪些调整！
- 例如，`AutomaticSpeechRecognitionPipeline` 具有 `chunk_length_s` 参数，
- 对于处理非常长的音频文件（例如，为整部电影或长达一小时的视频配字幕）非常有帮助，这通常是模型无法单独处理的：

## 在数据集上使用pipelines

### 这个部分是自制的数据集My example {i}，即一个字符串列表
- `pipelines`也可以对大型数据集进行推理。
- 建议使用迭代器来完成这一任务，这是最简单的方法：

```python
def data():
    for i in range(1000):
        yield f"My example {i}"


pipe = pipeline(model="gpt2", device=0)
generated_characters = 0
for out in pipe(data()):
    generated_characters += len(out[0]["generated_text"])
```

上面代码中

- 迭代器 `data()` 会产生每个结果
	- `f"My example {i}"` 是一个 f-string（格式化字符串），它会将变量 `i` 的当前值插入到字符串中。
	- 例如，当 `i` 为 0 时，生成的字符串是 `"My example 0"`；当 `i` 为 1 时，生成的字符串是 `"My example 1"`，以此类推。

- `pipelines`会自动识别输入为可迭代对象，并在GPU上处理数据的同时开始获取数据（在底层使用[DataLoader](https://pytorch.org/docs/stable/data.html#torch.utils.data.DataLoader)）。
	- `pipe(data())`：调用 `data()` 函数以获取生成器，并将其传递给文本生成管道。
	- 对于每个生成的输入（即 `"My example {i}"`），管道会生成相应的输出。
	- `out` 是管道返回的结果，通常是一个字典列表，其中包含生成的文本。
	- `out["generated_text"]`：访问第一个输出的生成文本（因为可能会有多个输出）。
	- `len(out["generated_text"])`：计算生成文本的字符长度，并将其累加到 `generated_characters` 中。

- `pipe(data())`分批次处理数据。这一点非常重要，因为您不必为整个数据集分配内存，可以尽可能快地将数据传送到GPU。

#### 减小模型尺寸：bitsandbytes
`bitsandbytes`可以传递8位加载的模型：
- 安装 `bitsandbytes` 
- 添加参数 `load_in_8bit=True`

- bitsandbytes是一个轻量级的Python库，
- 它可以让你在PyTorch中使用一些自定义的CUDA函数，特别是8位优化器、矩阵乘法（LLM.int8()）和量化函数。
- 它可以帮助你减小模型的尺寸，提高模型的性能和效率。
- 你需要先安装bitsandbytes软件包，方法是在终端中运行以下命令：

pip install bitsandbytes

- 可以使用load_in_8bit=True参数来指定你想要加载的模型是8位的。
- 例如，你可以使用Hugging Face的pipeline函数来创建一个生成器，它可以使用facebook/opt-1.3b模型来生成文本。
- 你只需要指定模型的名称和load_in_8bit=True，就可以让生成器在8位模式下运行，节省内存和时间。
- 以下是一个示例代码：

```python
# pip install accelerate bitsandbytes
import torch
from transformers import pipeline

pipe = pipeline(model="facebook/opt-1.3b", device_map="auto", model_kwargs={"load_in_8bit": True})
output = pipe("This is a cool example!", do_sample=True, top_p=0.95)
```

请注意，您可以将`checkpoint` 替换为任何支持大模型加载的Hugging Face模型，比如BLOOM！
